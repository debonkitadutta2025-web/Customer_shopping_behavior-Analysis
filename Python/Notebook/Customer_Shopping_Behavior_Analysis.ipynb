# 4. Data Cleaning

In this step, we clean and prepare the customer shopping dataset for
further analysis.

The cleaning process includes:

- Standardizing column names
- Removing unnecessary spaces
- Converting columns to appropriate data types
- Handling missing values
- Removing exact duplicate records
- Checking and handling invalid values
- Standardizing categorical values

The original dataset will not be overwritten. A separate cleaned dataset
will be created at the end of the project.

## 4.1 Create a Working Copy

Before making any changes, we create a copy of the original DataFrame.

This allows us to keep the original `df` unchanged and perform all
cleaning operations on a separate DataFrame named `clean_df`.

```python
# Create a copy of the original dataset

clean_df = df.copy()

print("Working copy created successfully!")
```

## 4.2 Standardize Column Names

The original column names contain spaces, brackets, and special
characters.

For example:

`Purchase Amount (USD)`

will be converted to:

`purchase_amount_usd`

Simple column names make Python, SQL, and Power BI analysis easier.

```python
# Clean and standardize column names

clean_df.columns = (
    clean_df.columns
    .str.lower()
    .str.strip()
    .str.replace(' ', '_')
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
    .str.replace('$', '', regex=False)
)

print("Cleaned column names:")
print(clean_df.columns.tolist())
```

## 4.3 Check the New Column Names

After standardizing the column names, we verify that all columns have
been renamed correctly.

The cleaned column names should be lowercase and use underscores
instead of spaces.

```python
# Display cleaned column names

for i, column in enumerate(clean_df.columns, start=1):
    print(i, "→", column)
```

## 4.4 Remove Extra Spaces from Text Columns

Text columns may contain unnecessary spaces before or after values.

For example:

` Male `

should become:

`Male`

We remove these unnecessary spaces from all text-based columns.

```python
# Remove leading and trailing spaces from text columns

text_columns = clean_df.select_dtypes(include='object').columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()

print("Extra spaces removed from text columns.")
```

## 4.5 Standardize Text Values

Categorical values should use a consistent format.

For example, gender values should be represented consistently as:

- Male
- Female

We will standardize the gender column using title case.

```python
# Standardize gender values

clean_df['gender'] = clean_df['gender'].str.title()

print(clean_df['gender'].unique())
```

## 4.6 Standardize Categorical Columns

We will remove unnecessary spaces from categorical columns and make sure
their values are stored consistently.

The categorical columns include product, category, location, size,
color, season, subscription status, shipping type, discount, promo code,
payment method, and purchase frequency.

```python
# Standardize categorical columns

categorical_columns = [
    'item_purchased',
    'category',
    'location',
    'size',
    'color',
    'season',
    'subscription_status',
    'shipping_type',
    'discount_applied',
    'promo_code_used',
    'payment_method',
    'frequency_of_purchases'
]

for col in categorical_columns:
    clean_df[col] = clean_df[col].astype(str).str.strip()

print("Categorical columns standardized.")
```

## 4.7 Convert Numerical Columns

The following columns should contain numerical values:

- Customer ID
- Age
- Purchase Amount
- Review Rating
- Previous Purchases

We convert these columns to numeric data types.

If an invalid value is encountered, `errors='coerce'` converts it into
a missing value (`NaN`) so that it can be detected and handled.

```python
# Convert numerical columns to numeric data types

numeric_columns = [
    'customer_id',
    'age',
    'purchase_amount_usd',
    'review_rating',
    'previous_purchases'
]

for col in numeric_columns:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors='coerce'
    )

print(clean_df[numeric_columns].dtypes)
```

## 4.8 Check Missing Values After Type Conversion

After converting numerical columns, some invalid values may have become
`NaN`.

Therefore, we check the dataset again for missing values.

```python
# Check missing values

missing_after_conversion = clean_df.isnull().sum()

print(missing_after_conversion)
```

## 4.9 Handle Missing Numerical Values

For numerical columns with missing values, we use the median value of
the respective column.

The median is useful because it is less affected by extreme values than
the mean.

```python
# Fill missing numerical values using the median

numeric_columns = [
    'customer_id',
    'age',
    'purchase_amount_usd',
    'review_rating',
    'previous_purchases'
]

for col in numeric_columns:
    if clean_df[col].isnull().sum() > 0:
        clean_df[col] = clean_df[col].fillna(
            clean_df[col].median()
        )

print("Missing numerical values handled.")
```

## 4.10 Handle Missing Categorical Values

For categorical columns, missing values are filled using the mode.

The mode represents the most frequently occurring value in a column.

```python
# Fill missing categorical values using the mode

categorical_columns = [
    'gender',
    'item_purchased',
    'category',
    'location',
    'size',
    'color',
    'season',
    'subscription_status',
    'shipping_type',
    'discount_applied',
    'promo_code_used',
    'payment_method',
    'frequency_of_purchases'
]

for col in categorical_columns:
    if clean_df[col].isnull().sum() > 0:
        clean_df[col] = clean_df[col].fillna(
            clean_df[col].mode()[0]
        )

print("Missing categorical values handled.")
```

## 4.11 Remove Exact Duplicate Rows

Exact duplicate rows represent completely identical records.

These records can cause transactions to be counted more than once.

We will remove exact duplicate rows from the working dataset.

```python
# Check duplicate rows before removal

duplicates_before = clean_df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)
```

```python
# Remove exact duplicate rows

clean_df = clean_df.drop_duplicates()

print("Duplicate rows removed.")
print("Current number of rows:", len(clean_df))
```

## 4.12 Handle Invalid Age Values

Age values outside a reasonable range should be investigated.

For this project, we consider ages below 18 or above 100 as invalid.

Instead of silently changing these values, we remove records with
invalid age values because they cannot be reliably corrected from the
available dataset.

```python
# Count invalid age records

invalid_age_count = (
    (clean_df['age'] < 18) |
    (clean_df['age'] > 100)
).sum()

print("Invalid age records:", invalid_age_count)
```

```python
# Remove invalid age records

clean_df = clean_df[
    (clean_df['age'] >= 18) &
    (clean_df['age'] <= 100)
].copy()

print("Invalid age records removed.")
```

## 4.13 Handle Invalid Purchase Amounts

Purchase amounts should be greater than zero.

Records with zero or negative purchase amounts are considered invalid
for this analysis.

We will remove such records if they exist.

```python
# Count invalid purchase amounts

invalid_purchase_count = (
    clean_df['purchase_amount_usd'] <= 0
).sum()

print(
    "Invalid purchase amount records:",
    invalid_purchase_count
)
```

```python
# Remove invalid purchase amounts

clean_df = clean_df[
    clean_df['purchase_amount_usd'] > 0
].copy()

print("Invalid purchase amounts handled.")
```

## 4.14 Handle Invalid Review Ratings

Review ratings should normally fall between 0 and 5.

Records outside this range are considered invalid and will be removed
from the cleaned dataset.

```python
# Count invalid review ratings

invalid_rating_count = (
    (clean_df['review_rating'] < 0) |
    (clean_df['review_rating'] > 5)
).sum()

print("Invalid rating records:", invalid_rating_count)
```

```python
# Remove invalid review ratings

clean_df = clean_df[
    (clean_df['review_rating'] >= 0) &
    (clean_df['review_rating'] <= 5)
].copy()

print("Invalid ratings handled.")
```

## 4.15 Handle Invalid Previous Purchases

Previous Purchases represents the number of previous purchases made by
the customer.

A negative number of purchases is not logically valid.

Therefore, records containing negative values will be removed.

```python
# Count invalid previous purchase values

invalid_previous_count = (
    clean_df['previous_purchases'] < 0
).sum()

print(
    "Invalid previous purchase records:",
    invalid_previous_count
)
```

```python
# Remove invalid previous purchase values

clean_df = clean_df[
    clean_df['previous_purchases'] >= 0
].copy()

print("Invalid previous purchase values handled.")
```

## 4.16 Final Missing Value Check

After handling missing values and invalid records, we perform another
missing-value check.

The cleaned dataset should not contain unexpected missing values.

```python
# Check missing values after cleaning

final_missing = clean_df.isnull().sum()

print(final_missing)
print(
    "\nTotal missing values:",
    final_missing.sum()
)
```

## 4.17 Final Duplicate Check

We check for duplicate rows one more time after completing the cleaning
operations.

```python
# Final duplicate check

print(
    "Duplicate rows after cleaning:",
    clean_df.duplicated().sum()
)
```

# 5. Feature Engineering

Feature engineering is the process of creating new meaningful features
from existing data.

In this project, we will create additional columns that help us better
understand customer shopping behavior.

The following features will be created:

1. **Age Group** — Groups customers into different age ranges.
2. **Customer Type** — Identifies customers as new or repeat customers.
3. **High Value Purchase** — Identifies purchases with an amount of
   $80 or more.

These features will make customer segmentation and business analysis
easier.

## 5.1 Create Age Group

Customers will be divided into five age groups:

- 18–25
- 26–35
- 36–45
- 46–55
- 56+

Grouping ages makes it easier to compare purchasing behavior across
different customer age groups.

```python
# Create age groups

clean_df['age_group'] = pd.cut(
    clean_df['age'],
    bins=[17, 25, 35, 45, 55, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56+']
)

print("Age group created successfully!")
```

```python
# Display age and age group

clean_df[['age', 'age_group']].head(10)
```

```python
# Count customers in each age group

clean_df['age_group'].value_counts().sort_index()
```

## 5.2 Create Customer Type

Customer type will be determined using the `previous_purchases` column.

Rules:

- If `previous_purchases` is 0 → **New Customer**
- If `previous_purchases` is greater than 0 → **Repeat Customer**

This feature helps us understand customer retention and repeat
purchasing behavior.

```python
# Create customer type

clean_df['customer_type'] = np.where(
    clean_df['previous_purchases'] > 0,
    'Repeat Customer',
    'New Customer'
)

print("Customer type created successfully!")
```

```python
# Display previous purchases and customer type

clean_df[
    ['previous_purchases', 'customer_type']
].head(10)
```

```python
# Count customer types

clean_df['customer_type'].value_counts()
```

## 5.3 Create High Value Purchase

We will create a new feature called `high_value_purchase`.

For this analysis:

- Purchase amount >= $80 → **High Value**
- Purchase amount < $80 → **Regular Value**

This feature will help us identify higher-value transactions and analyze
their customer, product, and category characteristics.

```python
# Create high-value purchase indicator

clean_df['high_value_purchase'] = np.where(
    clean_df['purchase_amount_usd'] >= 80,
    'High Value',
    'Regular Value'
)

print("High-value purchase feature created successfully!")
```

```python
# Display purchase amount and purchase value type

clean_df[
    ['purchase_amount_usd', 'high_value_purchase']
].head(10)
```

```python
# Count high-value and regular purchases

clean_df['high_value_purchase'].value_counts()
```
